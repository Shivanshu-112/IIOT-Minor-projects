```python
import numpy as np
```

```python
a=np.array([1,2,4])
b=np.array([2,7,9])
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)
```

```python
a=np.array([[5,4,9],[9,4,7],[5,2,8]])
b=np.array([[2,4,6],[9,8,7],[3,2,6]])
np.dot(a,b)
```

```python
a=np.array([[5,4,9],[9,4,7],[5,2,8]])
np.transpose(a)
```

```python
a=np.array([[5,4,9],[9,4,7],[5,2,8]])
b=np.array([[2,4,6],[9,8,7],[3,2,6]])
np.concatenate([a,b])
```
